**Question 4.** Take the average of the following set of numbers (formatted as a list).

```python
# Here's the set of numbers
data=[4.8, 10, 12.87, 3.6, 4, 16, 8.4, 3.14, 7, 6.442, 7.5, 15, 14.3]
avg = sum(data)/len(data)
avg
```

**Question 5.** Now take the standard deviation of the same set of numbers. The formula for standard deviation, $\sigma$, is: $\sigma = \sqrt{\frac{\Sigma (x_i - \mu)^2}{N-1}}$, where $\mu$ is the average of the data set and $N$ is the population size (number of numbers in the data set). The $\Sigma (x_i - \mu)^2$ means to take each number in the data set, subtract the data set average, square this value, and then sum all of this up; the $\Sigma$ is a summation sign.

**Note:** This is more involved than the previous questions, and there are many different ways to do this. You will probably need to use a loop for the summation part. If you find this question much harder than the previous ones that's ok, it is by design!

```python
a = 0
for i in data:
  b = (i-avg)**(2)
  a += b
((a)/(len(data)-1))**(1/2)
```

**Question 6.** Copy and paste the class height data into the code box below, and format it into a list.

```python
height = '''6-2
5-2
5-7
6-3
5-7
5-7
5-7
5-4
5-9'''
heights = height.split('\n')
heights
```

**Question 7.** Now you need to convert your list of strings into floats, so we can do some math with the heights as numbers. But you can't just convert straight away because of the way we formatted the numbers as feet and inches. Use the string manipulation commands (like strip or split, etc.) so that for each height you get two numbers, one for the feet and one for the inches. Then turn all of these into floats. There are multiple ways to do this, so just follow whatever makes the most sense to you.

```python
ft_in = []
for i in range(len(heights)):
  a = heights[i].split("-")
  ft_in.append(a)

floats = []
for i in range(len(ft_in)):
  floats.append(float(ft_in[i][0]))
  floats.append(float(ft_in[i][1]))
print(floats)
```

**Question 8.** Now that you have a list with floats giving the class height in feet and inches, calculate the average and standard deviation. But to do this, you will first want to convert all the heights into just inches. Then find the mean and standard deviations of class heights in inches, and then convert your answer for the mean back to feet and inches.

```python
ft = floats[0:17:2]
inc = floats[1:18:2]
inches = []

for i in range(len(ft)):
  inches.append(ft[i]*12 + inc[i])

mean = sum(inches)/len(inches)

t = 0
for i in inches:
  p = (i-mean)**(2)
  t += p
((t)/(len(inches)-1))**(1/2)
```
